```python
from roboflow import Roboflow
from getpass import getpass
from ultralytics import YOLO
import cv2
```

```python
api_key_input = getpass('Cole sua API Key do Roboflow aqui: ')

rf = Roboflow(api_key=api_key_input)
project = rf.workspace("roboflow-100").project("construction-safety-gsnvb")
version = project.version(2)
dataset = version.download("yolov8")

print("Dataset baixado com sucesso!")
```

```python
model = YOLO('yolov8n.pt')

results = model.train(
    data=f'{dataset.location}/data.yaml',
    epochs=30,
    imgsz=640,
    cache=True,
    save=True,
    device=0,
    patience=10,
    batch=16,
    name='yolo_ppe_v2'
)

print("Treinamento concluído")
```

```python
best_model = YOLO('runs/detect/yolo_ppe_v2/weights/best.pt')
metrics = best_model.val()

print(f"mAP50: {metrics.box.map50}")
print(f"mAP50-95: {metrics.box.map}")
```
